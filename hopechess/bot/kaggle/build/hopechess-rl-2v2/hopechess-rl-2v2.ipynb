# Hopechess RL 2v2
Self-play reinforcement learning for Hopechess (simultaneous-move chess), 1v1 and 2v2.
Code: hopechess/bot in ydiffraction/reinforcement-learning (packed below).

```python
import base64, io, tarfile, os, subprocess, sys
BLOB = ''
os.makedirs('/kaggle/working/bot', exist_ok=True)
tarfile.open(fileobj=io.BytesIO(base64.b64decode(BLOB)), mode='r:gz').extractall('/kaggle/working/bot')
os.chdir('/kaggle/working/bot')
r = subprocess.run('pip install -q pybind11', shell=True)
print(subprocess.run('bash build.sh', shell=True, capture_output=True, text=True))
print(subprocess.run('nvidia-smi; nproc; free -g', shell=True, capture_output=True, text=True).stdout)
```

```python
INIT = ''
```

```python
cmd = f'python3 -u train.py --out /kaggle/working/run {INIT} --hours 6 --parallel 192 --blocks 8 --channels 80 --k1 20 --k-single 12 --k-pairs 30 --iter-samples 30000 --bootstrap-games 300 --eval-every 5 --eval-games 40 --p2v2 1.0'
print(cmd)
p = subprocess.Popen(cmd, shell=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
with open('/kaggle/working/train.log', 'w') as log:
    for line in p.stdout:
        print(line, end='', flush=True); log.write(line); log.flush()
print('exit', p.wait())
```
